```python
import requests
import pandas as pd
```

```python
API_KEY = "b238da37213d7fa7c9edc37eec6312a5"
url = "https://api.openweathermap.org/data/2.5/weather"
cities = ["Lahore", "Karachi", "Islamabad", "London", "Tokyo"]
rows = []
for city in cities:
    params = {
        "q": city,
        "appid": API_KEY,
        "units": "metric"
    }
    res = requests.get(url, params=params)
    data = res.json()
    rows.append({
        "city": city,
        "temperature": data["main"]["temp"],
        "humidity": data["main"]["humidity"],
        "weather": data["weather"][0]["description"]
    })
```

```python
pd.DataFrame(rows)
```
