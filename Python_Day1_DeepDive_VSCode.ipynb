# PART 1 — VARIABLES, DEEPER

## 1.1 A variable is a labelled box

`=` does NOT mean "equals". It means **"put this value into this box"**.

````python
sales = 44680.60
print(sales)
````

## 1.2 Reassignment — the box can be refilled
There is only ever ONE box called `sales`. Putting something new in throws the old value away.

````python
sales = 1000
print(sales)

sales = 2000
print(sales)     # the 1000 is gone forever
````

## 1.3 Variables built from other variables

````python
price = 261.96
quantity = 2

total = price * quantity
print(total)
````

### The important difference from Excel
Change `price` now, then print `total` again.

````python
price = 500
print(total)     # STILL 523.92 — it did not update!
````

`total` was calculated **once** and stored. It is a photograph, not a live link.

In Excel a formula cell recalculates automatically. In Python you must run the calculation again.

````python
total = price * quantity     # recalculate
print(total)                # now it is 1000.0
````

## 1.4 Updating a variable using itself
This looks like broken maths, but read it **right to left**:
*calculate `count + 1` first, then store the answer back into `count`.*

````python
count = 0

count = count + 1
print(count)

count += 1          # shortcut for exactly the same thing
print(count)

count += 10
print(count)
````

The same shortcuts exist for the other operators.

````python
x = 100
x -= 20     # x = x - 20
x *= 2      # x = x * 2
x /= 4      # x = x / 4
print(x)
````

## 1.5 Two handy tricks

````python
# assign several at once
region, orders, revenue = "Central", 180, 44680.60
print(region, orders, revenue)
````

````python
# swap two values in one line
a, b = 10, 20
a, b = b, a
print(a, b)
````

**1. Using a variable that does not exist (or a typo)**

````python
# run this - it is meant to fail
print(total_sale)     # we called it total_sales, not total_sale
````

**2. Capital letters matter** — `Sales` and `sales` are two different boxes.

````python
sales = 100
print(Sales)     # meant to fail
````

**3. Never name a variable after a Python word.**
`print = 5` destroys the print function for the rest of your session.
(If you ever do this by accident: **Restart the kernel**.)

````python
# DO NOT do this in real code - shown once so you recognise it
# print = 5
# print("hello")   <- would now fail
print("Use names like total_sales, order_count, region_name")
````

# PART 2 — LISTS, DEEPER

**A list is a column of data.** Everything you do in pandas later is a supercharged version of this.

## 2.1 Creating lists

````python
sales   = [261.96, 731.94, 14.62]          # numbers
regions = ["Central", "West", "East"]      # text
mixed   = ["Central", 180, 44680.60]       # allowed, but avoid for real data
empty   = []                               # start empty, fill later

print(sales)
print(regions)
print(empty)
````

## 2.2 Indexing — counting starts at ZERO

```
  list:     [261.96, 731.94, 14.62, 957.58]
  index:        0       1       2       3
  negative:    -4      -3      -2      -1
```

````python
sales = [261.96, 731.94, 14.62, 957.58]

print(sales[0])     # first
print(sales[1])     # second
print(sales[-1])    # LAST - works no matter how long the list is
print(sales[-2])    # second from the end
````

`[-1]` is the most useful trick here — the last item, whether the list has 5 rows or 5 million.

## 2.3 Slicing — taking a piece of the list
This is the ancestor of `df.head()` in pandas.

**The end number is NOT included.**

````python
nums = [10, 20, 30, 40, 50, 60]

print(nums[0:3])    # [10, 20, 30]  - starts at 0, STOPS BEFORE 3
print(nums[:3])     # same thing, shorter
print(nums[3:])     # [40, 50, 60] - from 3 to the end
print(nums[-2:])    # [50, 60]     - last two
print(nums[:])      # the whole list
````

## 2.4 Changing a list

````python
sales = [261.96, 731.94, 14.62]
print(sales)

sales.append(957.58)        # add to the END
print(sales)

sales.insert(0, 100.00)     # add at a POSITION
print(sales)

sales.remove(14.62)         # remove by VALUE
print(sales)

last = sales.pop()          # remove the last one and hand it back
print(last, "was removed ->", sales)

sales[0] = 999.99           # overwrite by position
print(sales)
````

## 2.5 Reading a list without changing it

````python
nums = [30, 10, 50, 20]

print(len(nums))      # how many
print(sum(nums))      # total
print(max(nums))      # biggest
print(min(nums))      # smallest
````

### sorted() vs .sort() — an important difference

````python
nums = [30, 10, 50, 20]

print(sorted(nums))   # returns a NEW sorted list
print(nums)           # the original is untouched

nums.sort()           # changes the ORIGINAL permanently
print(nums)
````

````python
# biggest first
nums = [30, 10, 50, 20]
print(sorted(nums, reverse=True))
````

## 2.6 Searching a list

````python
regions = ["Central", "West", "East", "West"]

print(regions.count("West"))     # how many times it appears
print(regions.index("East"))     # what position it is at
print("South" in regions)        # is it there at all?
print("West" in regions)
````

> `in` is your `WHERE region IN (...)` from SQL. Same question, different language.

## 2.7 A realistic example — a mini sales summary

````python
sales = [261.96, 731.94, 14.62, 957.58, 22.37, 1706.18]

print(f"Rows:    {len(sales)}")
print(f"Total:   ${sum(sales):,.2f}")
print(f"Average: ${sum(sales)/len(sales):,.2f}")
print(f"Largest: ${max(sales):,.2f}")
print(f"Top 3:   {sorted(sales, reverse=True)[:3]}")
````

That last line is *"top 3 by sales"* — the same thing you wrote in SQL as
`ORDER BY sales DESC LIMIT 3`. Same idea, different tool.

# PART 3 — f-STRINGS, PROPERLY

## 3.1 What the `f` actually does
Without the `f`, Python prints the curly brackets literally.

````python
name = "Central"

print("Region: {name}")     # WRONG - prints the brackets
print(f"Region: {name}")    # RIGHT - the f makes {} live
````

`f` stands for **formatted**. It tells Python: *look inside the curly brackets and swap in the real value.*

## 3.2 Why it beats the old way

````python
name = "Central"
orders = 180

# the old way - ugly, and numbers need str()
print("Region: " + name + ", Orders: " + str(orders))

# the f-string way
print(f"Region: {name}, Orders: {orders}")
````

The `+` version **crashes** if you forget `str()`. f-strings do not care about type.

## 3.3 You can calculate inside the brackets

````python
total = 44680.60
orders = 180
name = "central"

print(f"Average: {total / orders}")
print(f"Double:  {total * 2}")
print(f"Region:  {name.upper()}")
````

## 3.4 Formatting — the part that matters for analysts

The pattern is `{value:format}`

````python
n = 44680.5977

print(f"{n}")           # raw
print(f"{n:.2f}")       # 2 decimal places
print(f"{n:.0f}")       # no decimals
print(f"{n:,}")         # thousands separator
print(f"{n:,.2f}")      # both  <- use this one for money
print(f"${n:,.2f}")     # with a currency sign
````

### Percentages — two ways

````python
margin = 0.0134

print(f"{margin:.2%}")        # the % multiplies by 100 for you
print(f"{margin*100:.2f}%")   # doing it manually - same answer
````

## 3.5 Alignment — making a report line up
- `:<10` left-align in 10 spaces
- `:>10` right-align in 10 spaces

````python
print(f"{'Central':<10} ${24990.19:>12,.2f}")
print(f"{'West':<10} ${11830.35:>12,.2f}")
print(f"{'East':<10} ${4496.16:>12,.2f}")
print(f"{'South':<10} ${3363.90:>12,.2f}")
````

Those are your real Superstore region totals — now printed as a tidy report.

## 3.6 The f-string cheat sheet

| You write | You get | Use it for |
|---|---|---|
| `{x}` | raw value | anything |
| `{x:.2f}` | `248.23` | decimals |
| `{x:,.2f}` | `44,680.60` | money |
| `{x:.1%}` | `1.3%` | percentages |
| `{x:<10}` | left-aligned | text columns |
| `{x:>10}` | right-aligned | number columns |

## 3.7 The error you will make
Forgetting the `f`. Run this and look closely at the output.

````python
region = "Central"
print("The region is {region}")     # no f - prints the brackets
print(f"The region is {region}")    # with f - correct
````
