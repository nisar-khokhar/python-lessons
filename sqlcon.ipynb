```python
cities = ["Lahore", "Karachi", "Islamabad"]
df_list = []

for city in cities:
    url = 'https://api.openweathermap.org/data/2.5/weather'
    params = {"q": city, "appid": API_KEY, "units": "metric"}
    res = requests.get(url, params=params)
    resp = res.json()
    df = pd.DataFrame([resp])
    df['temp'] = df['main'].str.get("temp")
    df['clarity'] = df["weather"].str[0].str.get("description")
    df['humidity'] = df["main"].str.get("humidity")
    df_list.append(df[["name", "temp", "clarity","humidity", "wind", "sys"]])


df_cities = pd.concat(df_list, ignore_index=True)
df_cities
```

```python
rows = []
for city in cities:
    url = 'https://api.openweathermap.org/data/2.5/weather'
    params = {"q": city, "appid": API_KEY, "units": "metric"}
    res = requests.get(url, params=params)
    resp = res.json()
    rows.append({
        "city": resp['name'],
        'temp': resp['main']['temp'],
        'humidity': resp['main']['humidity'],
        'clarity': resp['weather'][0]['description'],
    })

weather_df = pd.DataFrame(rows)
weather_df
    
```
