### 1.  Make an array
Create a NumPy array of these daily sales: 250, 800, 1200, 450, 999, 1500, 300, 2100. Call it `sales`.


```python
# your code here
sales = np.array([250, 800, 1200, 450, 999, 1500, 300, 2100])
```

### 2.  Inspect
Print the array's `shape`, `size`, and `dtype`.


```python
# your code here
print("Shape: ", sales.shape)
print("Size: ", sales.size)
print("dtype: ", sales.dtype)
```

### 3.  Generate
Use `np.arange` to make the even numbers from 0 to 20 (inclusive of 20 is fine or not — your call).


```python
# your code here
print(np.arange(0,20,2))
```

### 4.  Slicing
From `sales`, print the **first 3** values, and the **last** value.


```python
# your code here
print("First 3 values: ",sales[0:3])
print("Last Value: ", sales[-1])
```

### 5.  Vectorize
Apply a 10% discount to every value in `sales` (multiply by 0.9). No loop.


```python
# your code here
print(sales * 0.9)
```

### 6.  Boolean mask
How many days had sales over **1000**? Then show **which** values they were.


```python
# your code here
print("Number of days had sales over 1000: ", (sales > 1000).sum())
```

### 7.  Two conditions
Show sales that were **between 500 and 1500** (greater than 500 AND less than 1500).


```python
# your code here
sales[(sales>500) & (sales<1500)]
```

### 8.  np.where
Label each day `'target hit'` if sales > 1000, else `'below'`.


```python
# your code here
print(np.where(sales > 1000, "target hit", "below"))
```

### 9.  Reshape to 2D
Reshape `sales` (8 values) into a **2 x 4** grid. Call it `grid`.


```python
# your code here
sales = sales.reshape(2,4)
sales
```

### 10.  axis
On `grid`, compute the sum **down each column** (`axis=0`) and the sum **across each row** (`axis=1`).


```python
# your code here
print("axis = 0 : ", sales.sum(axis = 0))
print("axis = 1 : ", sales.sum(axis = 1))
```

### 11.  Broadcasting
Add the bonus `[10, 20, 30, 40]` to **every row** of `grid`.
