### 12.  New column
Make a column `outcome` = `'Survived'`/`'Died'` from `survived`, using `.map()`. Show 5 rows of `sex`, `class`, `outcome`.

### 13.  Age bands + groupby
Make an `age_group` column with `pd.cut` (Child 0-12, Teen 12-18, Adult 18-60, Senior 60+). Then show the **survival rate for each age group**. Which group did best?


```python
# your code here
# Define bins and labels
bins = [0, 12, 18, 60, 100]
labels = ['Child', 'Teen', 'Adult', 'Senior']

# Create the age_group column
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels, right=False)

# Calculate survival rate (as percentage) for each age group
survival_rates = df.groupby('age_group', observed=False)['survived'].mean() * 100

# Method B: Direct sum (adds up all the 1s)
survivor_counts = df.groupby('age_group', observed=False)['survived'].sum()

print(survivor_counts)
```
